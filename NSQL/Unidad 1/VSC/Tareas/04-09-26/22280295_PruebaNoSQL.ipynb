```python
# Importación de librerías
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Creación del DataFrame con los datos solicitados
datos = {
    'Modelo': ['Clave-valor', 'Documentos', 'Columnas', 'Grafos'],
    'Cantidad': [3, 5, 2, 4]
}

df = pd.DataFrame(datos)
df
```

```python
# Gráfica de barras con los datos del DataFrame
plt.figure(figsize=(6, 4))
plt.bar(df['Modelo'], df['Cantidad'], color=['#4C72B0', '#DD8452', '#55A868', '#C44E52'])
plt.title('Modelos de Bases de Datos NoSQL')
plt.xlabel('Modelo')
plt.ylabel('Cantidad')
plt.tight_layout()
plt.show()
```
