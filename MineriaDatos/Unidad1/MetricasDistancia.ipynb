Distancia Minkowsk

```python

import numpy as np
from itertools import combinations

# ---------------------------------------------------------
# Objetos
# ---------------------------------------------------------
objetos = {
    "O1": [5.8, 80, 20, 64,    1.78],
    "O2": [4.3, 83, 17, 68.89, 1.72],
    "O3": [7.0, 70, 10, 63,    1.75],
    "O4": [3.5, 65, 12, 57.2,  1.75],
    "O5": [2,   90, 25, 67.5,  1.80],
}


# ---------------------------------------------------------
# 1) DISTANCIA MINKOWSKI PASO A PASO
# ---------------------------------------------------------
def minkowski_distance_pasos(x, y, p, nombre_x, nombre_y):
    print(f"--- Distancia de Minkowski (p={p}) entre {nombre_x} y {nombre_y} ---")

    if p == 1:
        print("Formula: d(x,y) = sum |xi - yi|")
    elif p == 2:
        print("Formula: d(x,y) = raiz( sum |xi - yi|^2 )")
    else:
        print(f"Formula: d(x,y) = ( sum |xi - yi|^{p} )^(1/{p})")

    terminos = []
    for i in range(len(x)):
        diferencia = x[i] - y[i]
        valor_abs = abs(diferencia)
        potencia = valor_abs ** p
        terminos.append(potencia)
        print(f"  Paso {i+1}: |{x[i]} - {y[i]}| = |{diferencia:.4f}| = {valor_abs:.4f}   "
              f"->  {valor_abs:.4f}^{p} = {potencia:.4f}")

    suma_str = " + ".join(f"{t:.4f}" for t in terminos)
    suma = sum(terminos)
    print(f"\n  Suma de terminos: {suma_str} = {suma:.4f}")

    resultado = suma ** (1 / p)
    if p == 1:
        print(f"  Como p=1, el resultado es la suma directa: d = {resultado:.4f}")
    else:
        print(f"  Aplicando raiz {p}-esima: d = {suma:.4f}^(1/{p}) = {resultado:.4f}")

    print(f"\n  >>> Resultado final Minkowski p={p} ({nombre_x},{nombre_y}) = {resultado:.4f}\n")
    print("=" * 70)
    return resultado


```

Distancia Coceno

```python

# ---------------------------------------------------------
# 2) DISTANCIA COSENO PASO A PASO
# ---------------------------------------------------------
def distancia_coseno_pasos(x, y, nombre_x, nombre_y):
    print(f"--- Distancia Coseno entre {nombre_x} y {nombre_y} ---")
    print("Formula: distancia = 1 - ( (x . y) / (||x|| * ||y||) )")

    x = np.array(x, dtype=float)
    y = np.array(y, dtype=float)

    # --- Producto punto paso a paso ---
    print("\n  Paso 1: Producto punto (x . y)")
    productos = x * y
    suma_prod = 0.0
    for i in range(len(x)):
        suma_prod += productos[i]
        print(f"    {x[i]} * {y[i]} = {productos[i]:.4f}")
    print(f"    Suma del producto punto = {' + '.join(f'{p:.4f}' for p in productos)} = {suma_prod:.4f}")

    # --- Norma de x ---
    print("\n  Paso 2: Norma de x  ||x|| = raiz( sum xi^2 )")
    cuadrados_x = x ** 2
    for i in range(len(x)):
        print(f"    {x[i]}^2 = {cuadrados_x[i]:.4f}")
    suma_x2 = sum(cuadrados_x)
    print(f"    Suma = {' + '.join(f'{c:.4f}' for c in cuadrados_x)} = {suma_x2:.4f}")
    norma_x = np.sqrt(suma_x2)
    print(f"    ||x|| = raiz({suma_x2:.4f}) = {norma_x:.4f}")

    # --- Norma de y ---
    print("\n  Paso 3: Norma de y  ||y|| = raiz( sum yi^2 )")
    cuadrados_y = y ** 2
    for i in range(len(y)):
        print(f"    {y[i]}^2 = {cuadrados_y[i]:.4f}")
    suma_y2 = sum(cuadrados_y)
    print(f"    Suma = {' + '.join(f'{c:.4f}' for c in cuadrados_y)} = {suma_y2:.4f}")
    norma_y = np.sqrt(suma_y2)
    print(f"    ||y|| = raiz({suma_y2:.4f}) = {norma_y:.4f}")

    # --- Coseno de similitud ---
    print("\n  Paso 4: Coseno de similitud = (x . y) / (||x|| * ||y||)")
    denominador = norma_x * norma_y
    coseno_similitud = suma_prod / denominador
    print(f"    coseno = {suma_prod:.4f} / ({norma_x:.4f} * {norma_y:.4f})")
    print(f"    coseno = {suma_prod:.4f} / {denominador:.4f} = {coseno_similitud:.4f}")

    # --- Distancia coseno ---
    print("\n  Paso 5: Distancia coseno = 1 - coseno_similitud")
    distancia = 1 - coseno_similitud
    print(f"    distancia = 1 - {coseno_similitud:.4f} = {distancia:.4f}")

    print(f"\n  >>> Resultado final Distancia Coseno ({nombre_x},{nombre_y}) = {distancia:.4f}\n")
    print("=" * 70)
    return distancia


```
